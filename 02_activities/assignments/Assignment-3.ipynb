```python
# Import libraries

import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
import plotly as ply

np.random.seed(21)
```

```python
# pull up spreadsheet with the data
df = pd.read_excel('Water_Consumption_2024.xlsx')
print (df)
```

```python
# Define the variables

y1 = df['annual commercial usage']
y2 = df['annual residential usage']
x = df['city ward']
```

```python
# Create a figure and primary axes
fig, ax = plt.subplots(figsize=(10, 5))

# Plot the bars and label them
ax.bar(x,y1, label='Commercial')
ax.bar(x,y2, label='Residential')

# Add titles and labels
ax.set_title('2024 Residential versus Commercial Water Consumption by Ward in Toronto')
ax.set_ylabel('Water Usage in Cubic Metres')
ax.set_xlabel('City Ward')

# Add a legend
ax.legend(title='Account Type', loc='upper right', fontsize=12)

# Add horizontal grid lines
ax.grid(axis = 'y')

# List all City Wards
ax.xaxis.set_major_locator(plt.MultipleLocator(1))


```

I don't like how the Commercial value is hidden if it's lower than the Residential value and I could not find how to add one on top of the other, or how to display Residential & COmmercial side by side like the MacOS Numbers pre-made graph. So I made a simple histogram instead below

```python
# Create a figure and primary axes
fig, ax = plt.subplots(figsize=(15, 5))

# Plot the lines
ax.plot(x,y1)
ax.plot(x,y2)

# Add a title and labels
ax.set_title('2024 Residential versus Commercial Water Consumption by Ward in Toronto')
ax.set_ylabel('Water Usage in Cubic Metres')
ax.set_xlabel('City Ward')
ax.plot(x,y1, label = "Commercial", color='red')
ax.plot(x,y2, label = "Residential", color='blue')

# Add a legend
ax.legend(title='Account Type', loc='upper right', fontsize=12)

# Add grid lines
ax.grid(axis = 'y')

# List all the City Wards
ax.xaxis.set_major_locator(plt.MultipleLocator(1))
```
